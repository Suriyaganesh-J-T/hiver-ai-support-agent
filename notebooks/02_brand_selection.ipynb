```python
import pandas as pd

brand_stats = pd.read_csv(
    "../data/processed/brand_statistics.csv"
)

brand_stats.head(20)
```

```python
brand_stats.shape
```

```python
brand_stats.columns.tolist()
```

```python
candidates = [
    "AmazonHelp",
    "AppleSupport",
    "Uber_Support",
    "SpotifyCares",
    "AmericanAir",
    "Delta",
    "TMobileHelp",
    "VirginTrains",
    "SouthwestAir",
    "comcastcares",
]

candidate_stats = brand_stats[
    brand_stats["support_account"].isin(candidates)
].copy()

candidate_stats
```

```python
candidate_stats["customers_per_100_support_messages"] = (
    candidate_stats["unique_customers"]
    / candidate_stats["support_messages"]
    * 100
).round(2)

candidate_stats[
    [
        "support_account",
        "support_messages",
        "unique_customers",
        "customers_per_100_support_messages"
    ]
].sort_values(
    "unique_customers",
    ascending=False
)
```
